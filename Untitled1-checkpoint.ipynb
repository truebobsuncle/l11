```python
import os

filename = "my_test_file.txt"

# 1. Создаем и открываем файл, 2. Записываем в него несколько строк текста
with open(filename, "w", encoding="utf-8") as file:
    file.write("Первая строка текста.\n")
    file.write("Вторая строка текста.\n")
    file.write("Третья строка текста.\n")
# Файл автоматически закроется здесь благодаря оператору 'with' (Пункт 3)

# 3. Открываем файл снова, считываем содержимое и выводим на экран
print("--- Содержимое файла ---")
with open(filename, "r", encoding="utf-8") as file:
    content = file.read()
    print(content)
print("------------------------\n")

# 4. Проверяем, существует ли файл, используя os.path.exists()
if os.path.exists(filename):
    print(f"Проверка: Файл '{filename}' успешно найден на диске.")
else:
    print(f"Проверка: Файл '{filename}' не существует.")

# 5. Удаляем файл с помощью модуля os
os.remove(filename)
print(f"\nФайл '{filename}' был успешно удален.")

# Финальная проверка, чтобы убедиться в удалении
if not os.path.exists(filename):
    print("Подтверждение: Файла больше нет на диске.")
```
